```python

# PROGRAM PENYELESAIAN SPL 3 VARIABEL
# MENGGUNAKAN METODE ELIMINASI GAUSS

# 1. Memasukkan matriks koefisien dan konstanta
A = [
    [2, 3, 1],
    [1, 2, 3],
    [3, 1, 2]
]

B = [16, 17, 17]

n = len(B)

# 2. Membuat matriks augmented
M = []

for i in range(n):
    M.append(A[i][:] + [B[i]])

print("Matriks Augmented:")
for baris in M:
    print(baris)

# 3. Proses Eliminasi Gauss
for i in range(n):

    # Menentukan pivot terbesar
    pivot = i
    for j in range(i + 1, n):
        if abs(M[j][i]) > abs(M[pivot][i]):
            pivot = j

    # Menukar baris jika diperlukan
    M[i], M[pivot] = M[pivot], M[i]

    # Memeriksa apakah pivot bernilai nol
    if abs(M[i][i]) < 1e-12:
        raise ValueError("Sistem tidak memiliki solusi tunggal.")

    # Menghilangkan elemen di bawah pivot
    for j in range(i + 1, n):
        faktor = M[j][i] / M[i][i]

        for k in range(i, n + 1):
            M[j][k] = M[j][k] - faktor * M[i][k]

print("\nMatriks setelah Eliminasi Gauss:")
for baris in M:
    print([round(nilai, 4) for nilai in baris])

# 4. Proses Substitusi Mundur
X = [0.0] * n

for i in range(n - 1, -1, -1):
    jumlah = 0

    for j in range(i + 1, n):
        jumlah += M[i][j] * X[j]

    X[i] = (M[i][n] - jumlah) / M[i][i]

# 5. Menampilkan hasil perhitungan
print("\nHasil Penyelesaian:")
for i in range(n):
    print(f"{chr(120 + i)} = {X[i]:.4f}")

# 6. Memeriksa hasil dengan persamaan awal
print("\nPemeriksaan Hasil:")
for i in range(n):
    hasil = sum(A[i][j] * X[j] for j in range(n))
    print(
        f"Persamaan {i + 1}: "
        f"{hasil:.4f} = {B[i]}"
    )
```
